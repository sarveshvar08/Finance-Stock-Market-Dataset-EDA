## 4. Data Exploration

Before interpreting patterns, I checked missing values, duplicate records, and unusual values. This is important because missing values, duplicate records, and unusual values can affect summaries, calculation and visualization.


```python
# Missing values
missing_values = df.isnull().sum().sort_values(ascending=False)
print(missing_values)
```

```python
# Missing-value percentage
missing_summary = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (df.isnull().mean() * 100).round(2)
}).sort_values("Missing Count", ascending=False)

print(missing_summary)
```

```python
# Duplicate records
duplicate_count = df.duplicated().sum()
print("Duplicate rows:", duplicate_count)

# Check duplicate symbols
duplicate_symbols = df["Symbol"].duplicated().sum()
print("Duplicate Symbol values:", duplicate_symbols)
```

```python
# Unique-value counts for important categorical fields
for col in ["Listing Exchange", "Market Category", "ETF", "Test Issue", "Financial Status", "NextShares", "Nasdaq Traded"]:
    print(f"\n{col}:")
    print(df[col].value_counts(dropna=False))
```

```python
# Numerical summary
df[["Round Lot Size"]].describe()
```

### Data-quality summary

The dataset has **8,049 rows** and no missing values in several core identifier/category fields. Missing values are concentrated in `Financial Status` and `CQS Symbol`, so these columns should be handled carefully rather than automatically filling them with invented values.

The `Round Lot Size` field is numerical and can be summarized with descriptive statistics.

# 6. Required NumPy Demonstrations

The following cells demonstrate the required NumPy array creation, properties, indexing, slicing, reshaping, flattening, transpose, vectorization, broadcasting, aggregation, Boolean indexing, fancy indexing, and reproducible random-number generation.


```python
# Array creation using the required NumPy methods
ar_array = np.array([100, 200, 300, 400, 500])
ar_zeros = np.zeros(5)
ar_ones = np.ones(5)
ar_arange = np.arange(100, 501, 100)
ar_linspace = np.linspace(100, 500, 5)

print("np.array():", ar_array)
print("np.zeros():", ar_zeros)
print("np.ones():", ar_ones)
print("np.arange():", ar_arange)
print("np.linspace():", ar_linspace)
```

```python
# Array properties
finance_array = np.array(df["Round Lot Size"].to_numpy())

print("shape   :", finance_array.shape)
print("ndim    :", finance_array.ndim)
print("dtype   :", finance_array.dtype)
print("size    :", finance_array.size)
print("itemsize:", finance_array.itemsize, "bytes")
```

```python
# 1D indexing and slicing
print("First element:", arr_array[0])
print("Third element:", arr_array[2])
print("Slice [1:4]:", arr_array[1:4])

# 2D array indexing and slicing
two_d = np.arange(1, 13).reshape(3, 4)
print("\n2D array:\n", two_d)
print("Element at row 1, column 2:", two_d[1, 2])
print("First two rows:\n", two_d[0:2, :])
print("Columns 1 to 2:\n", two_d[:, 1:3])
```

```python
#  Demonstrating Reshape, flatten, ravel, and transpose
reshaped = np.arange(12).reshape(3, 4)

print("Original shape:", reshaped.shape)
print("Reshaped array:\n", reshaped)
print("\nflatten():", reshaped.flatten())
print("ravel():   ", reshaped.ravel())
print("\ntranspose():\n", reshaped.T)
```

### Practical meaning

- `reshape()` changes dimensions while keeping the same number of elements.
- `flatten()` converts an array to one dimension and generally returns a copy.
- `ravel()` also converts to one dimension and tries to return a view when possible.
- `transpose()` swaps rows and columns.

For example, a 3×4 array contains 12 elements, so a valid reshape could be 2×6 but not 3×5.


```python
# Vectorization: calculate a 2% hypothetical transaction adjustment
# without writing a loop for every element
sample_round_lots = df["Round Lot Size"].to_numpy()[:10]
adjusted_lots = sample_round_lots * 1.02

print("Original values :", sample_round_lots)
print("Adjusted values :", adjusted_lots)
```

```python
# Broadcasting: apply one common adjustment to every value
lot_values = np.array([100, 100, 100, 100, 100])
adjustment = 5

broadcast_result = lot_values + adjustment

print("Original:", lot_values)
print("Adjustment:", adjustment)
print("Result:", broadcast_result)
```

```python
# Aggregation functions
lot_values = df["Round Lot Size"].to_numpy()

print("Sum :", np.sum(lot_values))
print("Mean:", np.mean(lot_values))
print("Min :", np.min(lot_values))
print("Max :", np.max(lot_values))
print("Std :", np.std(lot_values))
print("Var :", np.var(lot_values))
```

```python
# Axis-based row-wise and column-wise operations
matrix = np.array([
    [100, 200, 300],
    [150, 250, 350],
    [120, 220, 320]
])

print("Matrix:\n", matrix)
print("\nColumn-wise mean (axis=0):", np.mean(matrix, axis=0))
print("Row-wise mean (axis=1):", np.mean(matrix, axis=1))
```

```python
# Boolean indexing
large_lot_values = lot_values[lot_values > 50]

print("Number of values greater than 50:", large_lot_values.size)
print("First 10 matching values:", large_lot_values[:10])
```

```python
# Fancy indexing
selected_positions = np.array([0, 10, 20, 30, 40])
selected_symbols = df["Symbol"].to_numpy()[selected_positions]

print("Selected positions:", selected_positions)
print("Selected symbols:", selected_symbols)
```

```python
# Reproducible random-number generation
np.random.seed(42)

simulated_returns = np.random.normal(loc=0, scale=0.02, size=10)

print("Simulated returns:")
print(np.round(simulated_returns, 4))
```

### Why use a fixed seed?

A fixed seed makes a random experiment reproducible. If the notebook is run again with the same seed and code, the same simulated values are generated. This is useful for demonstrations, testing, and verifying results.

# 7. Exploratory Visualisation

The visualizations focus on the categorical and numerical fields that actually exist in the dataset, because the selected file is securities metadata rather than historical price data

### Planned plots
1. Distribution of securities by **Listing Exchange**
2. Distribution of **ETF vs non-ETF** securities
3. Distribution of **Market Category**

A time-series closing-price chart is intentionally not included because the dataset contains no date or closing-price column.


```python
# Plot 1 — Listing Exchange distribution
exchange_counts = df["Listing Exchange"].value_counts()

plt.figure(figsize=(8, 5))
ax = sns.barplot(x=exchange_counts.index, y=exchange_counts.values)
plt.title("Number of Securities by Listing Exchange")
plt.xlabel("Listing Exchange")
plt.ylabel("Number of Securities")

for container in ax.containers:
    ax.bar_label(container, fmt="%d")

plt.tight_layout()
plt.show()
```

### Observations — Plot 1

- The securities are distributed across **five listing-exchange codes**.
- The distribution is not uniform; some exchange categories contain substantially more securities than others.
- This shows that the dataset has a strong categorical structure based on listing exchange.


```python
# Plot 2 — ETF status
etf_counts = df["ETF"].value_counts()

plt.figure(figsize=(7, 5))
ax = sns.barplot(x=etf_counts.index, y=etf_counts.values)
plt.title("ETF vs Non-ETF Securities")
plt.xlabel("ETF Flag")
plt.ylabel("Number of Securities")

for container in ax.containers:
    ax.bar_label(container, fmt="%d")

plt.tight_layout()
plt.show()
```

### Observations — Plot 2

- The dataset contains two ETF-status categories: **Y** and **N**.
- The non-ETF category is much larger than the ETF category.
- Therefore, most records in this file represent securities that are not marked as ETFs.


```python
# Plot 3 — Market Category distribution
market_counts = df["Market Category"].value_counts()

plt.figure(figsize=(8, 5))
ax = sns.barplot(x=market_counts.index, y=market_counts.values)
plt.title("Number of Securities by Market Category")
plt.xlabel("Market Category")
plt.ylabel("Number of Securities")

for container in ax.containers:
    ax.bar_label(container, fmt="%d")

plt.tight_layout()
plt.show()
```

### Observations — Plot 3

- The dataset contains **four market-category codes**.
- The categories have noticeably different frequencies, so market-category membership is not evenly distributed.
- This plot is useful for understanding the composition of the securities universe represented by the dataset.

# 8. Additional Numerical Check

The only continuous numerical field in this metadata file is `Round Lot Size`. Examining its distribution helps determine whether it provides useful variation for further EDA.


```python
# Numerical variable exploration — Round Lot Size
plt.figure(figsize=(8, 5))
sns.histplot(df["Round Lot Size"], bins=10, kde=False)
plt.title("Distribution of Round Lot Size")
plt.xlabel("Round Lot Size")
plt.ylabel("Number of Securities")
plt.tight_layout()
plt.show()

print(df["Round Lot Size"].value_counts().sort_index())
```

### Observation

The `Round Lot Size` variable has essentially no variation in this dataset: the observed value is 100 for all records. Therefore, although it is numerical, it does not provide much discriminatory information for comparing securities.

# 10. Limitations

### Limitation 1 — No historical price data
The dataset does not contain Date, Open, High, Low, Close, Adjusted Close, or Volume fields. Therefore, historical price and time-series analysis cannot be performed.

### Limitation 2 — Limited numerical variation
`Round Lot Size` is constant at 100 in the supplied data, so numerical statistical analysis is limited.

### Limitation 3 — Missing metadata
Some fields, especially `Financial Status` and `CQS Symbol`, contain missing values. These missing values may reduce the usefulness of those fields for certain comparisons.

### Limitation 4 — Association is not causation
The EDA identifies distributions and patterns in the available metadata. It does not prove that one variable causes another.
